```python

from tempfile import TemporaryDirectory
from langham_system import HotelSystem

def start_hotel_application():
    data_folder = TemporaryDirectory(prefix="langham_")
    system = HotelSystem(data_folder.name)
    print("LANGHAM HOTEL MANAGEMENT SYSTEM")
    print("Application initialized successfully!")
    return system, data_folder

hotel, workspace = start_hotel_application()
```

```python

# Menu Option 1 - Add New Rooms

room_details = [
    ("105", "Deluxe", "195"),
    ("210", "Family", "320"),
    ("305", "Suite", "450")
]

print("===== LANGHAM HOTEL - ROOM REGISTRATION =====")

for room_no, category, nightly_rate in room_details:
    message = hotel.add_room(room_no, category, nightly_rate)
    print(message)

print("===== ROOM REGISTRATION COMPLETED =====")
```

```python

# Menu Option 2 - Delete Room

print("===== LANGHAM HOTEL - ROOM DELETION =====")

room_number = "210"

print("Selected Room Number:", room_number)

deletion_result = hotel.delete_room(room_number)

print(deletion_result)

print("===== ROOM DELETION PROCESS FINISHED =====")
```

```python

# Menu Option 3 - Display Room Details

print("===== LANGHAM HOTEL - ROOM DETAILS =====")

print("Checking currently registered hotel rooms...")
print("-----------------------------------------")

room_information = hotel.display_rooms()

if room_information is not None:
    print(room_information)

print("-----------------------------------------")
print("===== ROOM DETAILS DISPLAY COMPLETED =====")

```

```python

# Menu Option 4 - Allocate Room

import inspect

print("===== LANGHAM HOTEL - NEW BOOKING =====")

guest_name = "Daniel Wilson"
room_id = "105"
stay_duration = "3"

print("Guest:", guest_name)
print("Room Number:", room_id)
print("Stay Duration:", stay_duration, "nights")

try:
    result = hotel.allocate(room_id, guest_name, stay_duration)
    print(result)
except (TypeError, ValueError) as error:
    print("Booking error:", error)
    print("Required parameters:", inspect.signature(hotel.allocate))

print("===== BOOKING PROCESS FINISHED =====")
```

```python

# Menu Option 6 - Billing and Room De-Allocation

import inspect

print("===== LANGHAM HOTEL - GUEST CHECKOUT =====")

checkout_room = "105"

print("Processing checkout for Room:", checkout_room)
print("-----------------------------------------")

try:
    final_bill = hotel.bill_and_release(checkout_room)

    if final_bill is not None:
        print(final_bill)

    print("-----------------------------------------")
    print("===== CHECKOUT PROCESS COMPLETED =====")

except (TypeError, ValueError) as error:
    print("Checkout error:", error)
    print("Required parameters:",
          inspect.signature(hotel.bill_and_release))
```

```python

# Menu Option 7 - Save Room Allocations to a File

student_id = "YOUR_STUDENT_ID"

print("===== LANGHAM HOTEL - SAVE BOOKINGS =====")

# Create a new booking
booking = hotel.allocate("305", "Sophia Brown", "2")
print(booking)

# Retrieve current allocations
booking_records = hotel.display_allocations()

# Save records to a text file
file_name = f"LHMS_{student_id}.txt"

with open(file_name, "w", encoding="utf-8") as file:
    file.write(booking_records or "No active bookings.")

print("File saved successfully:", file_name)
print("===== BOOKING RECORDS SAVED =====")
```

```python

# Menu Option 8 - Read Saved Room Allocations

print("===== LANGHAM HOTEL - SAVED BOOKING RECORDS =====")

print("Opening file:", file_name)
print("------------------------------------------")

try:
    with open(file_name, "r", encoding="utf-8") as booking_file:
        saved_data = booking_file.read()

    print(saved_data)
    print("------------------------------------------")
    print("Booking records retrieved successfully!")

except FileNotFoundError:
    print("Error: Booking file could not be found.")

print("===== FILE READING COMPLETED =====")
```

```python

# Menu Option 9 - Backup and Clear Booking Records

from datetime import datetime
from pathlib import Path

print("===== LANGHAM HOTEL - FILE BACKUP =====")

backup_time = datetime.now().strftime("%Y%m%d_%H%M%S")
backup_name = f"{Path(file_name).stem}_Backup_{backup_time}.txt"

try:
    # Read existing booking records
    with open(file_name, "r", encoding="utf-8") as file:
        records = file.read()

    if records.strip():
        # Append records to backup file
        with open(backup_name, "a", encoding="utf-8") as backup:
            backup.write(records + "\n")

        print("Backup created:", backup_name)

        # Clear original file after backup
        with open(file_name, "w", encoding="utf-8"):
            pass

        print("Original booking file cleared.")
        print("Backup completed successfully!")
    else:
        print("No booking records available to back up.")

except FileNotFoundError:
    print("Error: Original booking file not found.")

except OSError as error:
    print("File operation failed:", error)

print("===== BACKUP PROCESS FINISHED =====")
```

```python

# Task 3 - Built-in Functions and Modules

import ast
import builtins
import inspect
import langham_system

print("===== LANGHAM HOTEL - PYTHON FUNCTION ANALYSIS =====")

# Read and analyse the application source code
source_code = inspect.getsource(langham_system)
code_tree = ast.parse(source_code)

# Identify imported modules
imported_modules = set()

for node in ast.walk(code_tree):
    if isinstance(node, ast.Import):
        for item in node.names:
            imported_modules.add(item.name)

    elif isinstance(node, ast.ImportFrom):
        imported_modules.add(node.module or "")

print("\n1. IMPORTED MODULES")
for module in sorted(imported_modules):
    print("-", module)

# Identify built-in function calls
builtin_functions = set()

for node in ast.walk(code_tree):
    if isinstance(node, ast.Call):
        if isinstance(node.func, ast.Name):
            name = node.func.id

            if name in vars(builtins):
                builtin_functions.add(name)

print("\n2. BUILT-IN FUNCTIONS")
for function in sorted(builtin_functions):
    print("-", function)

print("\nTotal Modules:", len(imported_modules))
print("Total Built-in Functions:", len(builtin_functions))
print("===== FUNCTION ANALYSIS COMPLETED =====")
```
